### Data Cleaning

```python
df.drop(columns=['id','ca'],inplace=True) # drop unneccesary columns
```

```python
num_cols = ['trestbps','chol','thalch','oldpeak']  #numerical column - Median

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
cat_cols = ['fbs','restecg','exang','slope'] #Categorial columns - median

for col in cat_cols:
    df[col] =df[col].fillna(df[col].mode()[0])
```

```python
df['thal'] =df['thal'].fillna('Unknown')
```

```python
df.isnull().sum()
```

### Univariate Analysis - Study of One variable at a time

#### What is the age range?
#### Which age group is most common?

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'],bins=20,kde=True)
plt.title('Age Distibution')
plt.show()
```
